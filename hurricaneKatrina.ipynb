Use geeViz to quickly visualize landscape change using the Landscape Change Monitoring System (LCMS) data. 
- geeViz can be used to visualize and analyze any data in Google Earth Engine

Step 1: Initialize Earth Engine and import geeViz package for visualization

```python
import ee
# Initialize the Earth Engine module with a specific project
project = ''
ee.Initialize(project)
import geeViz.getImagesLib as gil
# Assing gil Map class as an object for visualization 
Map = gil.Map
```

Step 2. Import data and parameters and map data in geeViz

```python
Map.clearMap()
# Import Landscape Change Monitoring System (LCMS) data
lcms = ee.ImageCollection("USFS/GTAC/LCMS/v2024-10")

# Select Land Use
lu = lcms.select('Land_Use')

# Filter to years of interest
luYrs = lu.filter(ee.Filter.calendarRange(2004, 2010, 'year'))

# Map.addLayer(lu,{}, 'Land Use 2004-2023 time series')
Map.addTimeLapse(luYrs)
# Map.addTimeLapse(luYrs,{'min':0,'max':6}, 'Land Use 2004-2023 time lapse')

# Center map on New Orleans 9th ward
geom = ee.Geometry.Point(-89.9585206523721,30.04487803090007)
Map.centerObject(geom, 12)
Map.turnOnInspector()
Map.view()
```
